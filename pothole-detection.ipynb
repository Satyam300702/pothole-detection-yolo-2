```python
from ultralytics import YOLO
import os
import cv2
import random
import matplotlib.pyplot as plt
```

```python
dataset_path = "/kaggle/input/datasets/satyamkumar1330/pothole-dataset/pothole_dataset"
```

```python
train_images = dataset_path + "/train/images"
images_files = os.listdir(train_images)
sample = random.choice(images_files)

img = cv2.imread(os.path.join(train_images,sample))
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.figure(figsize=(6,6))
plt.imshow(img)
plt.axis("off")
```

```python
model = YOLO("yolo26n.pt")
```

```python
model.train( 
     data=f"{dataset_path}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=8,
    device = 0,


    mosaic=0.5,
    mixup=0.0,
    copy_paste=0.1,

    cos_lr=True,
    patience=20,
    save = True,
    save_period = 1,
    workers=2
    
)
```
